```python
import json
import time 
from tqdm import tqdm
import pandas as pd
import numpy as np 
import os 
```

```python
with open("image_depict_url.json", "r") as f:
    urls = json.load(f)
len(urls)
```

```python
img_path = '/Users/lidajie/Desktop/research/Dura_Generous_Interface/images/all_imgs_upload'
dura2025_path = '/Users/lidajie/Desktop/Dura2025'
field_imgs = f'{dura2025_path}/0627_depict_identification/field_images'
```

```python
field_urls = {}
```

```python
# for each image in each field, get their url info dict and organize into a json file to be retrieved in the interface

for dir in os.listdir(field_imgs):
    field_name = dir.split('_')[1]
    field_urls[field_name] = []

    subdir = f"{field_imgs}/{dir}"
    if not os.path.isdir(subdir): continue 

    for img in os.listdir(subdir):
        object_id = img.split('.')[0]
        if not object_id: continue 
        
        for d in urls:
            if d["object_id"] == object_id:
                field_urls[field_name].append(d)

    if len(field_urls) % 5 == 0:
        with open("field_url_0801.json", "w") as f:
            json.dump(field_urls, f, indent=2)
```

```python
# check if the number of files align with number of entries
filecnt = 0
for dir in os.listdir(field_imgs):
    subdir = f"{field_imgs}/{dir}"
    if not os.path.isdir(subdir): continue 

    for img in os.listdir(subdir):
        object_id = img.split('.')[0]
        if not object_id: continue 
        filecnt += 1

entrycnt = 0
for _, urllst in field_urls.items():
    for d in urllst:
        entrycnt += 1 

print(filecnt == entrycnt)
```
